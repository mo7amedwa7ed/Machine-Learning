## Logistic Regression From Scratch

#### import Libraries

```python
import numpy as np
```

### Build the Model

```python
class LogisticRegressionFromScratch :
    def __init__(self, learningRate: float = 0.01, nIterations: int =1000):
        self.lr = learningRate
        self.n_iterations = nIterations
        self.weights = None
        self.bais = None

    def _sigmoid(self, z: np.ndarray)-> np.ndarray :
        z  = np.clip(z,-500,500)
        return 1 /(1+ np.exp(-z))

    def fit(self, X:np.ndarray, y: np.ndarray) ->None:
        n_samples, n_features = X.shape
        y = y.reshape(-1,1)

        self.weights = np.zeros((n_features,1))
        self.bais = 0.0

        for _ in range(self.n_iterations):
            linearModel = np.dot(X, self.weights) + self.bais
            y_pred = self._sigmoid(linearModel)

            error = y_pred - y
            dw = (1 / n_samples) + np.dot(X.T, error)
            db = (1 / n_samples) + np.sum(error)

            self.weights -= self.lr * dw
            self.bais -= self.lr * db

    def predictProba(self, X: np.ndarray) -> np.ndarray:
        linearModel = np.dot(X, self.weights) + self.bais
        return self._sigmoid(linearModel)

    def predict(self, X: np.ndarray, threshold: float = 0.5) ->np.ndarray:
        probability = self.predictProba(X)
        return (probability >= threshold).astype(int) 

    def binaryCrossEntropy(self, y_true: np.ndarray, y_pred: np.ndarray) -> float:
        n_samples = len(y_pred)
        y_true = y_true.reshape((-1,1))
        y_pred = np.clip(y_pred, 1e-15, 1-1e-15)
        loss = -np.mean(
            y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
        )

        return float(loss)


    
```

### Test

```python
if __name__ == "__main__":
  np.random.seed(42)

  # Generate synthetic dataset
  X_train = np.random.randn(300, 2)
  # True rule: if 2*x1 - 1.5*x2 + 0.5 > 0, class 1 else 0
  z = X_train[:, 0] * 2.0 + X_train[:, 1] * -1.5 + 0.5
  y_train = (1 / (1 + np.exp(-z)) > 0.5).astype(int)

  # Initialize and train model
  model = LogisticRegressionFromScratch(learningRate=0.1, nIterations=1500)
  model.fit(X_train, y_train)

  # Evaluate predictions and accuracy
  predictions = model.predict(X_train)
  probabilities = model.predictProba(X_train)
  loss = model.binaryCrossEntropy(y_train, probabilities)
  accuracy = np.mean(predictions.ravel() == y_train)

  print(f"--- Learned Parameters ---")
  print(f"Learned Weights: {model.weights.ravel()}")
  print(f"Learned Bias:    {model.bais:.4f}")
  print(f"Final BCE Loss:  {loss:.4f}")
  print(f"Training Acc:    {accuracy * 100:.2f}%")
```
