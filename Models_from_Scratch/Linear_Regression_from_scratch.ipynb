## Linear Regression Implementation From Scratch

### Import Libraries

```python
import numpy as np
```

### Data

```python
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)
```

```python
print("X:",X)
print("Y:",y)
print("number of samples:",len(X))
```

```python
class LinearRegressionFromScratch:
    def __init__(self, learningRate: float =0.01, nIteration: int =1000):
        self.lr = learningRate
        self.n_iterations = nIteration
        self.weights = None
        self.bais = None

    def fit(self, X: np.ndarray, y: np.ndarray) -> None:
        n_samples, n_features = X.shape

        y = y.reshape(-1,1)

        self.weights = np.zeros((n_features,1))
        self.bais = 0.0

        for _ in range(self.n_iterations):
            y_pred = np.dot(X, self.weights) + self.bais
            error = y_pred - y

            dw = (2 / n_samples) * np.dot(X.T, error)
            db = (2 / n_samples) * np.sum(error)


            self.weights -= self.lr * dw
            self.bais -= self.lr * db

    def predict(self, X: np.ndarray) -> np.ndarray:
        return np.dot(X, self.weights) + self.bais


    def MSE(self, y_true: np.ndarray, y_pred:np.ndarray) ->float:
        return float(np.mean((y_true.reshape(-1,1) - y_pred)**2))    
        
```

### Test

```python
if __name__ == "__main__":
    # Generate synthetic data: y = 3*x1 + 2*x2 + 5
    np.random.seed(42)
    X_train = np.random.randn(100, 2)
    true_weights = np.array([[3.0], [2.0]])
    true_bias = 5.0

    # Target output with a bit of random noise
    noise = np.random.randn(100, 1) * 0.1
    y_train = np.dot(X_train, true_weights) + true_bias + noise

    # Initialize and train the model
    model = LinearRegressionFromScratch(learningRate=0.05, nIteration=1000)
    model.fit(X_train, y_train)

    # Predictions
    predictions = model.predict(X_train)
    mse = model.MSE(y_train, predictions)

    # Output results
    print(f"--- Learned Parameters ---")
    print(f"Learned Weights:\n{model.weights.ravel()}")
    print(f"Learned Bias:   {model.bais:.4f}")
    print(f"\nTarget Weights:  [3.0, 2.0]")
    print(f"Target Bias:     5.0")
    print(f"Training MSE:    {mse:.6f}")
```
