```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

```python
advertised = pd.read_csv("E:/Machine Learning/Datasets/tvmarketing.csv")
```

```python
advertised.plot(x='TV',y='Sales',kind='scatter')
```

```python
x = advertised[['TV']]
y = advertised['Sales']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.8,random_state=42)
model = LinearRegression()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

```python
y_pred[:10]
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test,y_pred)
r_squared = r2_score(y_test, y_pred)

print(f"Mean Absloute Error: {mae}")
print(f"Mean Square Error: {mse}")
print(f"R Squared: {r_squared}")
```

```python
c = range(1, len(y_test) + 1)
plt.plot(c,y_test , color='red',linewidth=2, linestyle="-")
plt.plot(c,y_pred , color='blue',linewidth=2, linestyle="-")
plt.title("Acutal & Predicted")
plt.xlabel('Index')
plt.ylabel('Sales')
plt.show()
```
